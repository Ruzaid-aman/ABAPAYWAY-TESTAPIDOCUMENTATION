```python
# Voice Design (English) - High Quality 1.7B Model
import torch
import soundfile as sf
from qwen_tts import Qwen3TTSModel
from IPython.display import Audio

# Load the Voice Design model
model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-0.6B-CustomVoice",
    device_map="cuda:0",
    dtype=torch.bfloat16,
    attn_implementation="sdpa",
)

# Generate voice
wavs, sr = model.generate_voice_design(
    text="This is a high-quality test using the one point seven billion parameter model.",
    language="English",
    instruct="Clear male voice, professional tone",
)

# Save and play the output
output_path = "output_voice_design_1.7b.wav"
sf.write(output_path, wavs[0], sr)
display(Audio(output_path))
```
